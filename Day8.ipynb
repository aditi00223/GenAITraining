```python
# client mostly ask each and every months \
# give me 10 pandas dataset like id , name , post, salary, city and age 

# 1. add columns bonus > = 40000  5% 
# 2. ------- hra 10% 
# 3. -------- da 5 % 
# 4. calculate gross salary 
# 5. calculate net salary 
import pandas as pd

# Employee dataset
data = {
    "id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "name": ["Rahul", "Neha", "Vikram", "Pooja", "Rohan",
             "Sneha", "Arjun", "Kavya", "Varun", "Megha"],
    "post": ["Manager", "Developer", "HR", "Tester", "Developer",
             "Designer", "Manager", "Developer", "Tester", "HR"],
    "salary": [60000, 50000, 45000, 38000, 55000,
               42000, 70000, 48000, 35000, 46000],
    "city": ["Delhi", "Mumbai", "Pune", "Jaipur", "Bangalore",
             "Chennai", "Hyderabad", "Pune", "Kolkata", "Delhi"],
    "age": [32, 25, 29, 24, 27, 26, 35, 23, 28, 30]
}

df = pd.DataFrame(data)

# 1. Bonus = 5% if salary >= 40000
df["bonus"] = df["salary"].apply(
    lambda x: x * 0.05 if x >= 40000 else 0
)

# 2. HRA = 10% of salary
df["hra"] = df["salary"] * 0.10

# 3. DA = 5% of salary
df["da"] = df["salary"] * 0.05

# 4. Gross Salary
df["gross_salary"] = (
    df["salary"] + df["bonus"] + df["hra"] + df["da"]
)

# 5. Net Salary
df["net_salary"] = df["gross_salary"]

print(df)
```

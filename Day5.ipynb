```python
class Test:
    # class level variable
    a = 10
    def __init__(self):
        # instance level variable
        self.b = 20

t = Test()
print(t.a, t.b)
t.a = 11
t.b = 22
print(t.a, t.b)

t1= Test()
print(t1.a,t1.b) # a,b instance variable 
# instance variable is varied from object to object 

```

```python
class Demo:
    def msg(self):
        print("bye")
    @classmethod
    def show(cls):
        print("hello")
    @staticmethod
    def disp():
        print("hi")
d= Demo()
# we can access instance method by using object reference
d.msg()
# we can classmethod by using object reference as well as class name
Demo.show()
d.show()
#w e can access static method by using only class name 
Demo.disp()
```

```python
# diff b/w instance vs static/class vs local variable 
class Student:
    college = "ABC University"  # class variable 

    def __init__(self, name, rollNo):
        self.name = name        # instance variable 
        self.rollNo = rollNo    # instance variable

    def show(self):
        greeting = "Hello, " + self.name  # local variable
        print(greeting)
        print(self.college, self.name, self.rollNo)


s1 = Student("Ravi", 101)
s2 = Student("Simran", 102)

s1.show()
s2.show()
```

```python
# Inheritance - the process of aquiring properties and behaviour from one class to another class is known as inheritance 
# there are 5 types of inheritance 
# 1 single level inheritance - when a class inherits the properties the properties of another class is known as 
# 2 multi level - when a class inherits the properties of another nut these class also inherited another class
# 3 multiple inheritance - when a class inherits from more than one class
# 4 Hierarchical - when a class inherited by more than one class
# 5 hybrid - it is a combination of heirarchical and multiple inheritance 
```

```python
class A:
    def m1(self):
        print("hello")
    def m2(self, a, b):
        self.a = a      # store as instance variable
        self.b = b      # store as instance variable
        print(self.a + self.b)

class B(A):
    pass

b = B()
b.m1()          # works fine — inherited from A
b.m2(5, 10)     # now works — passes required args, prints 15
```

```python
class A:
    def m1(self):
        print("hi")
    def m2(self):
        print("hello")

class B(A):
    def m3(self):
        print("bye")
class C(B):
    pass

c = C()
c.m1()
c.m2()
c.m3()
```

```python
import numpy as np

sales = np.array([120, 150, 180, 90, 200, 250, 175, 300, 220, 160])
print( sales.size)
print(sales.ndim)
print( sales.dtype)
print( sales.shape)
```

```python

total_sales = np.sum(sales)

print( total_sales)
```

```python

average_sales = np.mean(sales)

print( average_sales)
```

```python

maximum_sales = np.max(sales)
minimum_sales = np.min(sales)
index_max = np.argmax(sales)
index_min = np.argmin(sales)

print( maximum_sales)
print(minimum_sales)
print( index_max)
print(index_min)
```

```python

print( sales[0])
print(sales[-1])
print(sales[4])
print( sales[2:7])
```

```python

reverse_sales = sales[::-1]

print("Reversed Sales Array:", reverse_sales)
```

```python

sales_greater_200 = sales[sales > 200]

print("Sales greater than 200:", sales_greater_200)
```

```python
import numpy as np

ages = np.array([22, 35, 28, 41, 19, 32, 25, 45, 30, 27])
```

```python

ages_between = ages[(ages >= 25) & (ages <= 35)]

print("Ages between 25 and 35:", ages_between)
```

```python
marks = np.array([
    [85, 78, 92],
    [67, 75, 80],
    [90, 88, 95],
    [55, 60, 65],
    [72, 85, 78]
])
```

```python

print("First student's marks:", marks[0])
print("Second student's marks:", marks[1])
print("Marks in first subject:", marks[:, 0])
print("Marks in second subject:", marks[:, 1])
```

```python

subject_average = np.mean(marks, axis=0)

print( subject_average)
```

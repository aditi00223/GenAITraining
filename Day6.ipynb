```python
from abc import ABC,abstractmethod
class Demo(ABC):
    @abstractmethod
    def show():
        pass
class Test(Demo):
    def show(self):
        print("hello")
t = Test()
t.show()

#d = Demo()
    
```

```python
class Test1(ABC):
    def __init__(self):
        self.a = 10
        self.b = 20
    @abstractmethod
    def show():
        pass
class Test2(Test1):
    def show():
        print(self.a)
        print(self.b)
        
        
```

```python
class Test1(ABC):
    @abstractmethod
    def m1():
        pass
class Test2(ABC):
    @abstractmethod
    def m2():
        pass
class Test(Test1,Test2):
    def m1(self):
        print("m1")
    def m2(self):
        print("m2")
t = Test()
t.m1()
t.m2()
```
